```python
import pandas as pd

```

```python
name = input("Enter student name: ")
math = float(input("Enter Math mark: "))
english = float(input("Enter English mark: "))
computer = float(input("Enter Computer mark: "))
```

```python
average = (math + english + computer) / 3

print("\n--- STUDENT RESULT ---")
print("Name:", name)
print("Average:", round(average, 2))
```

```python
if average >= 70:
    grade = "A"
elif average >= 60:
    grade = "B"
elif average >= 50:
    grade = "C"
elif average >= 40:
    grade = "D"
else:
    grade = "F"

print("Grade:", grade)
```

```python
if average >= 50:
    result = "PASS"
else:
    result = "FAIL"

print("Result:", result)
```
